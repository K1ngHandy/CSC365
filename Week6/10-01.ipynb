```python
from collections import Counter
import numpy as np

documents = [
    "data science is fun",
    "data science is powerful",
    "machine learning is fun",
]

tokenized_docs = [doc.split() for doc in documents]
vocab = sorted(set(word for doc in tokenized_docs for word in doc))
print("Vocabulary:", vocab)
dt_matrix = np.zeros((len(documents), len(vocab)))

for i, doc in enumerate(tokenized_docs):
    counts = Counter(doc)
    for j, term in enumerate(vocab):
        dt_matrix[i, j] = counts[term]

print("\nRaw Count Matrix:\n, dt_matrix")
tf_matrix = dt_matrix / dt_matrix.sum(axis = 1, keepdims = True)
print("\nTF Matrix:\n, tf_matrix")
```

```python
import pandas as pd
from textblob import TextBlob

tweets = pd.DataFrame({
    "tweet":[
        "I love Amazon",
        "Python is too hard sometimes...",
        "Big Data of the AI is the fantastic future",
    ]
})

def get_sentiment(text):
    analysis = TextBlob(text)
    if analysis.sentiment.polarity > 0:
        return "Positive"
    elif analysis.sentiment.polarity < 0:
        return "Negative"
    else:
        return "Neutral"
tweets['sentiment'] = tweets['tweet'].apply(get_sentiment)
print(tweets)

import matplotlib.pyplot as plt
sentiment_counts = tweets['sentiment'].value_counts()
sentiment_counts.plot(kind = 'pie')
plt.title("Sentiment Distribution of Tweets")
plt.ylabel("")
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

df = pd.DataFrame({
    "msg":[
        "teach me to call",
        "Call me a cab",
        "please call me.... PLEASE 44!",
        "please do not call me"],
        'label':[0, 1, 0, 1]
})

X_train, X_test, y_train, y_test = train_test_split(df['msg'], df['label'])
vect = CountVectorizer()
train_dtm = vect.fit_transform(X_train)
print("Document-term matrix shape:", train_dtm.shape)
print("Feature names:", vect.get_feature_names_out())

test_simple = ["please teach me to call a cab"]
test_simple_dtm = vect.transform(test_simple)
print(test_simple_dtm.toarray())
pd.DataFrame(test_simple_dtm.toarray(), columns = vect.get_feature_names_out())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB #simple text classifier

df = pd.DataFrame({
    'msg':[
        "call you tonight",
        "Call me a cab",
        "please call me... PLEASE 44!",
        "please do not call me",
    ],
    'label':[0, 1, 0, 1]
})

X_train, X_test, y_train, y_test = train_test_split(df['msg'], df['label'])
vect = CountVectorizer()
train_dtm = vect.fit_transform(X_train)
test_dtm = vect.transform(X_test)
model = MultinomialNB()
model.fit(train_dtm, y_train)
y_pred = model.predict(test_dtm)
print("Predicted labels for test messages:", y_pred)
new_msg = ["please tonight call me"]
new_msg_dtm = vect.transform(new_msg)
new_pred = model.predict(new_msg_dtm)
print("Prediction for new message:", new_pred)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

df = pd.read_csv("headlines.csv")
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(df['text'])
num_clusters = 3
kmeans = KMeans(n_clusters = num_clusters)
kmeans.fit(X)
df['cluster'] = kmeans.labels_
for i in range(num_clusters):
    print(f"\nCluster {i}:")
    cluster_texts = df[df['cluster'] == i]['text']
    for text in cluster_texts:
        print(f" - {text}")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")

vectorizer = TfidfVectorizer(max_features=100)
X_train = vectorizer.fit_transform(train_df["text"])
X_test = vectorizer.transform(test_df["text"])
model = MultinomialNB()
model.fit(X_train, train_df["label"])
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(test_df["label"], y_pred))
print(classification_report(test_df["label"], y_pred))

```
